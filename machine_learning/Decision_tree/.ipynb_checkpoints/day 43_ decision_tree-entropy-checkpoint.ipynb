```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
```

# Binary Classification

```python
df = pd.read_csv("../../datasets/student_binary_classification1.csv")
df
```

```python
# encoding 

df["Result"]=df["Result"].map({
    "Fail" : 0,
    "Good" : 1,
    "Average" : 2,
    "Excellent" : 3
})
df
```

```python
# features selection

X = df[["StudyHours"]]
y = df["Result"]
```

```python
# train_test_split
```

```python
model = DecisionTreeClassifier(
    criterion = "gini",
    random_state= 42
)
```

```python
# model train 

model.fit(X, y)
```

```python
print(model.tree_.threshold)
```

```python
new_student = pd.DataFrame({
    "StudyHours": [2,3,5]
})

predicition = model.predict(new_student)
print(predicition)
```

# multiple feature multi Classification

```python
df = pd.read_csv("../../datasets/student_multiclass_multiple_features.csv")
df
```

```python
# encoding 

df["Result"]=df["Result"].map({
    "Fail" : 0,
    "Average" : 1,
    "Good" : 2
})
df
```

```python
# features selection

X = df[["StudyHours", "Attendance", "PreviousMarks"]]
y = df["Result"]
```

```python
model = DecisionTreeClassifier(
    criterion = "gini",
    random_state= 42
)
```

```python
# model train 

model.fit(X, y)
```

```python
new_student = pd.DataFrame({
    "StudyHours":[5],
    "Attendance":[60],
    "PreviousMarks": [70]
    
})

model.predict(new_student)
```

# Multi Label

```python
df = pd.read_csv("../../datasets/student_multilabel.csv")
df
```

```python
X = df[
    [
        "StudyHours",
        "EnglishScore",
        "MathScore"
    ]
]
y = df[
    [
        "PassEnglish",
        "PassMath"
    ]
]
```

```python
model = DecisionTreeClassifier(
    criterion = "gini",
    random_state= 42
)
```

```python
model.fit(X, y)
```

```python
new_studentss = pd.DataFrame({
    "StudyHours":[5],
    "EnglishScore":[80],
    "MathScore": [65]
})
model.predict(new_studentss)
```
