```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("datasets/student_marks.csv")
# df.info()
# df.describe()
# df.isnull().sum()
```

```python
# features k ho (input) , hami lai chaiyako label(output/predicted value) k ho 

X = df[["StudyHours"]]   # --> df[["StudyHours"]] - one dimensional data ||| to convert to two dimensional data --> df[["StudyHours"]] - two dimensional data
Y = df["FinalMarks"]   # X is features(input) and Y is Label

# plot on scattter graph to have a look on data
plt.scatter(X,Y)
plt.xlabel("Study Hours")
plt.ylabel("Final Marks")
plt.title("Study Hours Vs Final Marks")
```

```python
# convert to numpy array 
X = X.to_numpy()
Y = Y.to_numpy()
```

```python
# split array for training and test array
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size = 0.2, random_state = 42)   # pass features, label, test_size = 0.2 (20%)
X_train      # every time print new data , to prevent from this we pass random_state=42, which fix the output
```

```python
# machine Learning

# training model using training data
model = LinearRegression()
model.fit(X_train, Y_train)
```

```python
# model verification using testing data

predicted_Y = model.predict(X_test)
predicted_Y
```

```python
# comparision

comparision = pd.DataFrame({
    "Actual" : Y_test,
    "Predicted" : predicted_Y
})
comparision
```

```python
new_students = np.array([
    [2],
    [4],
    [6],
    [8],
    [10]
])

marks_prediction = model.predict(new_students)
marks_prediction
```
