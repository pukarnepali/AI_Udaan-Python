```python
from sklearn.metrics import (confusion_matrix, accuracy_score, precision_score, recall_score, f1_score)
```

```python
# actual result 

y_test = [1,0,0,1]

# predicted result 
y_predictios = [1,0,1,0]

cm = confusion_matrix(y_test, y_predictios)

tn,fp,fn,tp = cm.ravel()

print("True Negative :", tn)
print("False Positive :", fp)
print("False Negative :", fn)
print("True Positive :", tp)
```

```python
print(accuracy_score(y_test, y_predictios))
print(precision_score(y_test, y_predictios))
print(recall_score(y_test, y_predictios))
print(f1_score(y_test, y_predictios))
```
