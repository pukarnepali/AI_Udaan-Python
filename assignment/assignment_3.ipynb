```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dp = pd.read_csv("../datasets/dp_lms_student_analytics_dataset.csv")
dp
```

```python
# line chart 
# monthly students enrollments

# convert Enrollmentdate to datetime
dp["EnrollmentDate"] = pd.to_datetime(dp["EnrollmentDate"])

# count enrollments by month
monthly_enrollment = ( dp.groupby(dp["EnrollmentDate"].dt.to_period("M")).size())

# plot line chart

plt.plot(monthly_enrollment.index.astype(str),
        monthly_enrollment.values,
        marker ="o")

plt.xlabel("Months")
plt.ylabel("Number of Enrollments")
plt.title("Monthly Enrollments of Students")
plt.grid()
# plt.figure(figsize=(10,4))
```

```python
# Monthly revenue trend
monthly_revenue = (dp.groupby(dp["EnrollmentDate"].dt.to_period("M"))["Revenue"].sum())

# line chart
plt.plot(monthly_revenue.index.astype(str), monthly_revenue.values, marker ="o")
plt.xlabel("Months")
plt.ylabel("Revenue")
plt.title("Monthly Revenue Trend")
plt.grid()
```
