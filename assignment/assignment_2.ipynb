```python
# standarize string values
for column in string_columns:
    lms[column] = lms[column].str.strip().str.title()

lms
```

```python
# detect neagtive value from all numeric columns
numeric_column = ['student_id','age','course_fee','discount','assignment_score','final_score','rating','attendance_percent']

for column in numeric_column:
    print(f"{column} : {(lms[column] < 0).sum()} negative value")
```

```python
# detect negative value from specific column
lms[lms["course_fee"] <0 ]
```

```python
# replace value 
lms.loc[lms["course_fee"]<0,"course_fee"] = 500
```

```python
# detect negative value from specific column
lms[lms["assignment_score"] <0 ]
```

```python
# replace value 
lms.loc[lms["assignment_score"] <0, "assignment_score"]= 5
```

```python
# outliers by IQR in "course_fee" Column

cQ1 = lms["course_fee"].quantile(0.25)
cQ1
```

```python
cQ3 = lms["course_fee"].quantile(0.75)
cQ3
```

```python
cIQR = cQ3-cQ1
cIQR
```

```python
clower_bound = cQ1-1.5*cIQR
clower_bound
```

```python
cupper_bound = cQ3+1.5*cIQR
cupper_bound
```

```python
c_outlier = lms[(lms["course_fee"] < clower_bound)|(lms["course_fee"] > cupper_bound)]
c_outlier
```

```python
# fix outliers by clip()
lms["course_fee"] = lms["course_fee"].clip(clower_bound,cupper_bound)
```

```python
# outliers by IQR in "attendence_percent" Column

aQ1 = lms["attendance_percent"].quantile(0.25)
aQ1
```

```python
aQ3 = lms["attendance_percent"].quantile(0.75)
aQ3
```

```python
aIQR = aQ3-aQ1
aIQR
```

```python
alower_bound = aQ1 - 1.5*aIQR
alower_bound
```

```python
aupper_bound = aQ3 + 1.5*aIQR
aupper_bound
```

```python
a_outlier = lms[(lms["attendance_percent"]<alower_bound)|(lms["attendance_percent"]>aupper_bound)]
a_outlier
```

```python
# outliers by IQR "assignment_score" Column

sQ1 = lms["assignment_score"].quantile(0.25)
sQ3 = lms["assignment_score"].quantile(0.75)
sIQR = sQ3-sQ1
slower_bound = sQ1-1.5*sIQR
supper_bound = sQ3+1.5*sIQR
s_outlier = lms[(lms["assignment_score"]< slower_bound)|(lms["assignment_score"]>supper_bound)]
s_outlier
```

```python
# outliers by IQR "final_score" Column

fQ1 = lms["final_score"].quantile(0.25)
fQ3 = lms["final_score"].quantile(0.75)
fIQR = fQ3 - fQ1
flower_bound = fQ1 - 1.5*fIQR
fupper_bound = fQ3 + 1.5*fIQR
f_outlier = lms[(lms["final_score"]<flower_bound)|(lms["final_score"]>fupper_bound)]
f_outlier
```

```python
# outliers by z_score "course_fee"

c_mean = lms["course_fee"].mean()
c_mean
```

```python
# standard deviation

c_std = lms["course_fee"].std()
c_std
```

```python
lms["c_zscore"] = (lms["course_fee"]-c_mean)/c_std
```

```python
c_outliers = lms[(lms["c_zscore"]>3)|(lms["c_zscore"]<-3)]
c_outliers
```

```python
# outliers by z_score "attendance_percent"

a_mean = lms["attendance_percent"].mean()
a_std = lms["attendance_percent"].std()
lms["a_zscore"] = (lms["attendance_percent"]-a_mean)/a_std
a_outliers = lms[(lms["a_zscore"] >3)|(lms["a_zscore"]< -3)]
a_outliers
```

```python
# outliers by z_score "assignment_score"

s_median = lms["assignment_score"].median()
s_mean = lms["assignment_score"].mean()
s_std = lms["assignment_score"].std()
lms["s_zscore"] = (lms["assignment_score"]-s_mean)/s_std
s_outliers = lms[(lms["s_zscore"] >3)|(lms["s_zscore"]< -3)]
s_outliers
```

```python
# fix outliers with median
lms.loc[(lms["s_zscore"] >3)|(lms["s_zscore"]<-3),"assignment_score"] = s_median
```

```python
# outliers by z_score "final_score"

f_mean = lms["final_score"].mean()
f_std = lms["final_score"].std()
lms["f_zscore"] = (lms["final_score"]-f_mean)/f_std
f_outliers = lms[(lms["f_zscore"] >3)|(lms["f_zscore"]< -3)]
f_outliers
```
