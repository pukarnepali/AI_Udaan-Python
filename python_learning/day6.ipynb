# Function in python

even can make code more managable and simple.


#syntax of function 

def function_name():
    block of code/statement

```python
names = ["Pukar", "Aakash", "Pradip"]

def greet():
    print("Hello", i)

for i in names:
    greet()
```

argument : data / value / input injection

parameter : receive garne placeholder

```python
def greet(namee):           # name is parameter
    print("Hello", namee)

greet("ankit")              # "ankit" is argument
```

```python
def sum(a,b):
    print("sum :", a+b)

sum(1,2)
sum(90,50)
```

```python
# return keyword 

def sum(a,b):
    return a+b

add = sum(1,4)      # return throws value into some variable , like in this line, 5 is captured in add variable
print(add)
```

```python
def student():
    return "Pukar", 25, "Pokhara"

student()       # returns in tuple data type

name,age,location = student()
print(age)      # specific valuee print
```

```python
# dafault value in function 

def say_hello(name = "default"):
    print("hallo", name)

say_hello()
```

```python
def multiple(num1, num2):
    print(num1, num2)
    print(num1*num2)

multiple(2,3)
multiple(num2=2, num1=5)
```
