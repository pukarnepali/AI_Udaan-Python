```python
age = int(input("Enter your age: "))

if age > 18:
    print("you can vote")
else:
    print("you cannot vote")
```

```python
password = input("Enter your password: ")

if password == "Nepal123":
    print("LOGIN")
else:
    print("password incorrect")
```

```python
i = int(input("Enter your number: "))

if i ==0:
    print("zero")
elif i < 0:
    print("negative")
elif i % 2 ==0:
    print("even")
else:
    print("odd")
```

```python
# Login system --> email, password

# method1 (use of nested if)
email= "email@gmail1.com"
password = "email123"

if email == "email@gmail.com":
    if password == "email123":
        print("login")
else:
    print("failed")
```

```python
# method2 (use of and operator)

email= "email@gmail.com"
password = "email123"

if email == "email@gmail.com" and password == "email123":
        print("login")
else:
    print("failed")
```

# Logical operator 

and operator = output is true if all condition is true 
or operator = output is true if one of the condition is true 
not operator = output is true if condition is false

```python
logged_in = False

if not logged_in:
    print("please login")
```

1. for loop 

DRY => Don't repeat yourself 

syntax:

for iterator in range(argument):

```python
#print 0 to 4 using for loop 
for i in range(5):
    print(i, end=",")
```

```python
#print 5 to 14 using for loop 
for i in range(5,15):
    print(i, end=" ")
```

```python
#print 0 to 20 but only even number using for loop 
for i in range(0,21, 2):
    print(i, end=" ")
```

```python
for i in range(11,0, -1):
    print(i, end=" ")
```

```python
countries = ["japan", "nepal", "usa"]

for i in countries:
    print(i)
```

```python
predicition_score = [77,98,23,86]

for i in predicition_score:
    if i > 80:
        print(i, "good score")
    else:
        print(i, "bad score")
```

```python
email_list=[
    "discount",
    "free ticket",
    "sale sale",
    "project update",
    "congrats"
]

for email in email_list:
    if "discount" in email or "congrats" in email or "free" in email:
        print("spam:", email)
    else:
        print("not spam:", email)


# in = checks whether specific string is in statement or not 
```
