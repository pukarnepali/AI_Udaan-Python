```python
#collections datatype / non primitive datatype
#list datatype
name =["manish", "akash", "anish"]
print(name)
print(name[0])  #print using indexing

name.append("prakash")  #add items to the list 
print(name)

name.pop()  #remove last item of the list
print(name)

name.insert(0, "mandip")    #add item according to the indexing
print(name)

name.remove("mandip")   #remove specific item from list 
print(name)

name[1]= "ankit"    #replace item of the list
print(name)

print(len(name))    #length of the list (items count)
```

```python
#sets datatype
# no duplicate value, unordered
country = {"nepal", "germany", "canada", "nepal"}

print(country)
```

```python
#tuple datatype
#unchangeable value
fruits = ("apple", "banana", "orange")
print(fruits)

print(fruits[1])
print(fruits.count("apple"))    # count specific items of tuple
print(fruits.index("orange"))   # find out index of specific item
```

```python
#dictionary
person_details = {
    "full_name" : "akash nepali",
    "age" : 25,
    "country" : "nepal",
    "isnepali" : True,
    "height" : 5.67
}

print(person_details["full_name"])      # accessing specific value from its key

person_details["color"] = "blue"       # add key and value in dictionary

person_details["age"] =26        # replace value in dictionary

person_details.pop("color")     # remove key and value from dictionary

print(person_details)
```
