```python
import matplotlib.pyplot as plt
import pandas as pd 
```

```python
bp1 = pd.DataFrame({
    "Employee_Name" : ["Ram", "Shyam", "Haari", "Ramesh", "Dinesh", "Bikash", "Ankit", "Rankit"],
    "Salary" : [1000,2000,3000,4000,5000,6000,7000,8000]
})
```

```python
# box plot 

plt.boxplot(bp1["Salary"])
plt.title("Employee Salary")
plt.ylabel("Salary")
```

```python
bp2 = pd.DataFrame({
    "Employee_Name" : ["Ram", "Shyam", "Haari", "Ramesh", "Dinesh", "Bikash", "Ankit", "Rankit","hmm"],
    "Salary" : [1000,2000,3000,4000,5000,6000,7000,8000,20000]
})
```

```python
# box plot with outliers

plt.boxplot(bp2["Salary"])
```
