```python
# importing matplotlib in jupyter lab 
# import matplotlib as mp
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# use pandas to retrive datasets
dd = pd.read_csv("../datasets/daraz_dataset.csv")
dd
```

```python
category = dd.groupby("Category")["Product"].count()
category
```

```python
# use customized color for pie chart
colored = [
    "green",
    "brown",
    "black",
    "blue",
    "red",
    "purple",
    "pink"
]
# pie chart using matplotlib
# shadow (boolean value for shadow of pie chart)
# startangle ( turn pie chart)
# explode ( for highlight specific product, it takes tuple)
plt.pie(category.values, labels=category.index, colors=colored, shadow=True, startangle=90, explode=(0.1,0,0,0,0,0,0.1))

# size of pie chart
# plt.figure(figsize=(6,6))

# pie chart's title 
plt.title("Product Distribution by category", fontsize = 20, color = "maroon")
```

```python
gender_products = dd.groupby("Gender")["Product"].count()
gender_products
```

```python
plt.pie(gender_products.values, labels=gender_products.index, colors = ["pink", "blue"], startangle=90)
plt.title("Product by Gender")
```

```python
# histogram 
# color (color of histogram)
# aplha (transparency)
# edge color (for edge color)
# bins ( default is 8, can be modified as per request)
plt.hist(dd["UnitPrice"], color="maroon", edgecolor = "blue", alpha= 0.3, bins = 20)

plt.title("Distribution of Unit Price")

#xlable and ylabel for name of x and y axis 
plt.xlabel("Unit Price")
plt.ylabel("Frequency")
```

```python
# customize bins and bins interval

bins = [0,5000,10000,20000,50000,80000]

plt.hist(dd["UnitPrice"], color="maroon", edgecolor = "blue", alpha= 0.3, bins = bins)
plt.title("Distribution of Unit Price")

#xlable and ylabel for name of x and y axis 
plt.xlabel("Unit Price")
plt.ylabel("Frequency")
```

```python
# Scatter plot 

data ={
    "Age" : [20,21,23,27,29,32],
    "TotalAmount" :[1000,1200,1500,2000,3000,9000]
}
dfd = pd.DataFrame(data)
plt.scatter(dfd["Age"], dfd["TotalAmount"], color = "green", s = 8)
plt.xlabel("Age")
plt.ylabel("Total Amount")
plt.title("Age vs Total Amount")
plt.grid(True)
```

```python
plt.scatter(dd["UnitPrice"], dd["Rating"])
plt.xlabel("Unit Price")
plt.ylabel("Rating")
plt.title("Unit Price vs Rating")
plt.grid(True)
```
