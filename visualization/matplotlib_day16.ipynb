```python
index = [1,2,3,4]    #x-label
price = [1000,2000,3000,4300]    #y-label

# figure is use to decrease or increase the size of canvas, default is (6,6)
plt.figure(figsize =(6,6))

# use plot() for line graph 
plt.plot(index, price)
plt.show()   # used to shoe line graph in old editor , not useful in jupyter lab 

# use bar() for bar graph
# plt.bar(index,price)
```

```python
# use pandas to retrive datasets
dd = pd.read_csv("../datasets/daraz_dataset.csv")
dd
```

```python
category = dd.groupby("Category")["Product"].count()
category
# category.index  # categories
# category.values   # product number
```

```python
# bar() for bar garph 
plt.figure(figsize = (18,8))  # increasing size of bar graph from default size
plt.bar(category.index, category.values, color = "maroon", edgecolor = "blue")  # change color of bars and change edge color 
plt.xlabel("Categories", fontsize = 20)  # x label name 
plt.ylabel("Products", fontsize = 20)  # y label name 
plt.title("Products and category Bar Graph")  # title
plt.grid(axis = "y", linestyle = "--", alpha = 0.35)   # grid in bar garph , linestyle to provide style on grid , aplha to visibility of grid
```
