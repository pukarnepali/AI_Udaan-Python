```python
age = int(input("Enter your age: "))

if age > 18:
    print("you can vote")
else:
    print("you cannot vote")
```

```python
i = int(input("Enter your number: "))

if i ==0:
    print("zero")
elif i < 0:
    print("negative")
elif i % 2 ==0:
    print("even")
else:
    print("odd")
```

```python
# Login system --> email, password

# method1 (use of nested if)
email= "email@gmail1.com"
password = "email123"

if email == "email@gmail.com":
    if password == "email123":
        print("login")
else:
    print("failed")
```

```python
# method2 (use of and operator)

email= "email@gmail.com"
password = "email123"

if email == "email@gmail.com" and password == "email123":
        print("login")
else:
    print("failed")
```

# Logical operator 

and operator = output is true if all condition is true 
or operator = output is true if one of the condition is true 
not operator = output is true if condition is false

```python
logged_in = False

if not logged_in:
    print("please login")
```
