```python
# convert to numeric for multiple columns

numeric_columns = [
    "Age",
    "Quantity",
    "UnitPrice",
    "DiscountPercent",
    "TotalAmount",
    "Rating"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors= "coerce")
```

```python
# # checking null value 
df.isnull().sum()

# # to fill null value/missing value
df["Age"] = df["Age"].fillna(df["Age"].mean())
```
