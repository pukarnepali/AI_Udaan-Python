```python
import numpy as np
```

```python
#vectorization & broadcasting
number = np.array([1,2,3,4,5])
results = number * 10
print(results > 30)
print(results > 30)
print(results)
results[(results > 30) & (results < 50)]
```

```python
#vectorization
price = np.array([100,200,300])
quantity = np.array([2,4,6])

price * quantity
```
