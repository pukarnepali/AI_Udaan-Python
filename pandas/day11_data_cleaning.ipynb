```python
import pandas as pd
```

```python
bdd = pd.read_csv("../datasets/bad_daraz_dataset.csv")
bdd
```

```python
# bdd.isnull()  # detect null values , missing values
bdd.isnull().sum()   # sum of null values from each column 
```

```python
# bdd.fillna(0)  # detect and fill missing values with zero
```

```python
# bdd["price"].fillna(bdd["price"].mean())    # returns error cause string is in price column
bdd["price"] = pd.to_numeric(bdd["price"], errors = "coerce")  # convert string value into null value with no error
bdd
bdd["price"] = bdd["price"].fillna(bdd["price"].mean())  # filling in mean value in null value
bdd
```

```python
bdd["age"] = pd.to_numeric(bdd["age"], errors ="coerce")
bdd["age"] = bdd["age"].fillna(bdd["age"].mean())
bdd

# bdd.dropna()  #drop/delete rows with null value
```

```python
bdd["rating"]=pd.to_numeric(bdd["rating"], errors = "coerce")
bdd
bdd["rating"] = bdd["rating"].fillna(bdd["rating"].mean())
bdd["rating"]
```

```python
# bdd.duplicated()  # detect duplicated data

bdd[bdd.duplicated()]
```

```python
# bdd["discount_percent"] = bdd["discount_percent"].str.replace("%","", regex=False)  # replace % from column
bdd
bdd["discount_percent"] = pd.to_numeric(bdd["discount_percent"], errors= "coerce")
bdd
bdd["discount_percent"] = bdd["discount_percent"].fillna(bdd["discount_percent"].mean())
bdd
```
