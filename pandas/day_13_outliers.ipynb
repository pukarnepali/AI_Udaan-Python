```python
import pandas as pd 
```

```python
sd = pd.DataFrame({

    "Salary": [1000,2000,3000,5000,7000,8000,9000,12000, 26000]
})
sd
```

```python
median = sd["Salary"].quantile(0.5)
median
```

```python
q1 = sd["Salary"].quantile(0.25)
q1
```

```python
q3 = sd["Salary"].quantile(0.75)
q3
```

```python
#IQR
iqr = q3-q1
iqr
```

```python
lower_bound = q1-(1.5*iqr)
lower_bound
```

```python
upper_bound = q3+(1.5*iqr)
upper_bound
```

```python
outliers = sd[(sd["Salary"]< lower_bound)|(sd["Salary"]> upper_bound)]
outliers
```
